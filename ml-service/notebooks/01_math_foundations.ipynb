# 01 — Mathematical Foundations

Linear algebra (SVD pada feature matrix wallet), calculus (gradient descent dari nol untuk logistic regression), probability (Bayes untuk klasifikasi naive).

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
```

## 1. Linear Algebra — SVD pada Feature Matrix

PCA via SVD: $X = U \Sigma V^T$. Simulasi feature matrix 50 wallet x 5 fitur.

```python
X = np.random.randn(50, 5) * [[1, 3, 0.5, 2, 0.2]] + [[5, 0, -2, 1, 3]]
X_centered = X - X.mean(axis=0)
U, S, Vt = np.linalg.svd(X_centered, full_matrices=False)
print(f'Singular values: {S.round(3)}')
print(f'Explained variance ratio: {(S**2 / (S**2).sum()).round(3)}')
# Project ke 2D
X_2d = U[:, :2] @ np.diag(S[:2])
plt.scatter(X_2d[:, 0], X_2d[:, 1], alpha=0.7)
plt.title('SVD projection to 2D'); plt.xlabel('PC1'); plt.ylabel('PC2'); plt.show()
```

## 2. Calculus — Gradient Descent untuk Logistic Regression

Binary classification wallet: bot (1) vs human (0). Loss = binary cross-entropy.

```python
def sigmoid(z): return 1/(1+np.exp(-z))

# synthetic data
n_samples = 200
X_bin = np.c_[np.random.randn(n_samples, 2), np.ones(n_samples)]
true_w = np.array([0.8, -0.5, 0.1])
y_bin = (sigmoid(X_bin @ true_w + 0.3*np.random.randn(n_samples)) > 0.5).astype(float)

w = np.random.randn(3) * 0.1
lr, losses = 0.1, []
for _ in range(500):
    pred = sigmoid(X_bin @ w)
    loss = -np.mean(y_bin*np.log(pred+1e-10) + (1-y_bin)*np.log(1-pred+1e-10))
    grad = X_bin.T @ (pred - y_bin) / n_samples
    w -= lr * grad
    losses.append(loss)

plt.plot(losses); plt.title('GD Convergence'); plt.xlabel('Iter'); plt.ylabel('Log Loss')
plt.show()
print(f'Final w: {w.round(3)}, true w: {true_w}')
```

## 3. Probability — Bayes untuk Klasifikasi Naive

$P(class|features) \propto P(features|class) \cdot P(class)$

Simulasi: apakah wallet dengan dex_swap > 0.3 dan tx_count > 100 adalah trader?

```python
# synthetic: P(trader)=0.3, P(dex>0.3|trader)=0.8, P(tx>100|trader)=0.6
# P(dex>0.3|not_trader)=0.1, P(tx>100|not_trader)=0.2
p_trader = 0.3
likelihood_trader = 0.8 * 0.6  # P(evidence|trader)
likelihood_not = 0.1 * 0.2       # P(evidence|not_trader)
evidence = likelihood_trader * p_trader + likelihood_not * (1-p_trader)
posterior = likelihood_trader * p_trader / evidence
print(f'P(trader | dex>0.3, tx>100) = {posterior:.3f}')
```
