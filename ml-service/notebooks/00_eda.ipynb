# 00 — Exploratory Data Analysis (EDA)

Distribusi fitur wallet, korelasi, missing values, outlier detection, visualisasi awal.

Tujuan: memahami shape & characteristics dataset sebelum modeling.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style='whitegrid')
%matplotlib inline
```

```python
# Simulasi dataset (ganti dengan load dari Postgres / parquet)
np.random.seed(42)
n = 1000
df = pd.DataFrame({
    'tx_count': np.random.lognormal(3, 1.5, n).astype(int),
    'active_days': np.random.lognormal(2, 1, n).astype(int),
    'native_in': np.random.exponential(10, n),
    'native_out': np.random.exponential(8, n),
    'gas_spent': np.random.exponential(0.5, n),
    'dex_swap_ratio': np.random.beta(2, 8, n),
    'failed_tx_ratio': np.random.beta(1, 20, n),
    'unique_counterparties': np.random.lognormal(1, 1, n).astype(int),
    'contract_interaction_ratio': np.random.beta(3, 3, n),
    'avg_tx_value': np.random.exponential(2, n),
})
print(f'{len(df)} wallets, {len(df.columns)} features')
df.head(3)
```

```python
# Descriptive statistics
df.describe(percentiles=[.01, .05, .25, .5, .75, .95, .99]).T.style.background_gradient(axis=1)
```

```python
# Missing values
missing = df.isnull().sum()
print('Missing per column:')
print(missing[missing > 0] if missing.sum() > 0 else 'None')
```

```python
# Distribusi fitur numerik
fig, axes = plt.subplots(3, 4, figsize=(16, 10))
for ax, col in zip(axes.flat, df.columns[:12]):
    ax.hist(df[col].clip(df[col].quantile(0.001), df[col].quantile(0.999)), bins=50, alpha=0.7)
    ax.set_title(col, fontsize=9); ax.tick_params(labelsize=7)
plt.tight_layout(); plt.show()
```

```python
# Correlation matrix
corr = df.corr()
mask = np.triu(np.ones_like(corr, dtype=bool), k=1)
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r', center=0, ax=ax,
            linewidths=0.5, annot_kws={'size':7})
ax.set_title('Feature Correlation Matrix'); plt.show()
```

```python
# Outlier detection — IQR per fitur
Q1, Q3 = df.quantile(0.25), df.quantile(0.75)
IQR = Q3 - Q1
outlier_counts = ((df < (Q1 - 1.5*IQR)) | (df > (Q3 + 1.5*IQR))).sum()
print('Outliers per feature (>1.5*IQR):')
print(outlier_counts[outlier_counts > 0].sort_values(ascending=False))
```
