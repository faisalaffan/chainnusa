# 06 — Time Series Analysis

Forecast aktivitas wallet (daily tx count) dengan ARIMA + Prophet.
Mendeteksi tren jangka panjang dan musiman.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import warnings; warnings.filterwarnings('ignore')
```

```python
# Synthetic daily tx count
np.random.seed(42)
days = 180
trend = np.linspace(5, 20, days)
weekly = 3 * np.sin(2 * np.pi * np.arange(days) / 7)
noise = np.random.normal(0, 2, days)
tx_count = np.maximum(trend + weekly + noise, 0).astype(int)
ts = pd.Series(tx_count, index=pd.date_range('2024-01-01', periods=days, freq='D'))

fig, ax = plt.subplots(figsize=(14, 4))
ts.plot(ax=ax); ax.set_title('Daily Transaction Count'); plt.show()
```

```python
# ACF & PACF
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
plot_acf(ts, lags=40, ax=ax1); plot_pacf(ts, lags=40, ax=ax2); plt.show()
```

```python
# ARIMA forecast
train, test = ts[:150], ts[150:]
model = ARIMA(train, order=(3, 1, 2))
fitted = model.fit()
print(fitted.summary().tables[0])
```

```python
forecast = fitted.forecast(len(test))
fig, ax = plt.subplots(figsize=(14, 4))
train.plot(ax=ax, label='Train'); test.plot(ax=ax, label='Test')
pd.Series(forecast, index=test.index).plot(ax=ax, label='ARIMA Forecast', linestyle='--')
ax.legend(); ax.set_title('ARIMA(3,1,2) Forecast'); plt.show()
```
