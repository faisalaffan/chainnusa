# 05 — Statistical Inference

Hypothesis testing: apakah avg gas wallet bot > human?
CLT demonstration, sampling distributions, confidence intervals.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from scipy import stats
sns.set_theme(style='whitegrid')
```

## Hypothesis Test: Gas Spending — Bot vs Human

H₀: µ_bot = µ_human (no difference)
H₁: µ_bot > µ_human (bot spend more gas)

```python
np.random.seed(42)
gas_human = np.random.exponential(0.2, 200)  # avg 0.2 ETH
gas_bot = np.random.exponential(0.5, 80)     # avg 0.5 ETH

t_stat, p_val = stats.ttest_ind(gas_bot, gas_human, alternative='greater')
print(f'Human gas mean={gas_human.mean():.3f}, std={gas_human.std():.3f}')
print(f'Bot gas mean={gas_bot.mean():.3f}, std={gas_bot.std():.3f}')
print(f't-statistic={t_stat:.3f}, p-value={p_val:.5f}')
print(f'Significant at α=0.05: {p_val < 0.05}')
```

```python
# Visual comparison
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.hist(gas_human, bins=30, alpha=0.6, label='Human')
ax1.hist(gas_bot, bins=30, alpha=0.6, label='Bot')
ax1.legend(); ax1.set_title('Gas Spending Distribution')
stats.probplot(gas_bot, dist='expon', plot=ax2)
ax2.set_title('Q-Q Plot (Bot gas vs Exponential)'); plt.show()
```

## Central Limit Theorem Demonstration

Distribution of sample means → normal (regardless of original distribution).

```python
# Highly skewed data (exponential)
population = np.random.exponential(10, 100000)
sample_means = [np.mean(np.random.choice(population, 100)) for _ in range(1000)]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.hist(population, bins=100, alpha=0.7); ax1.set_title('Population (skewed)')
ax2.hist(sample_means, bins=40, alpha=0.7); ax2.set_title('Sample Means (n=100, ~normal)')
plt.show()
print(f'Population skewness: {stats.skew(population):.2f}')
print(f'Sample means skewness: {stats.skew(sample_means):.2f}')
```

## Confidence Interval: avg daily tx per wallet

```python
daily_tx = np.random.lognormal(1, 0.8, 150)
mean, std = daily_tx.mean(), daily_tx.std()
ci = stats.t.interval(0.95, df=len(daily_tx)-1, loc=mean, scale=std/np.sqrt(len(daily_tx)))
print(f'Mean daily tx: {mean:.2f}')
print(f'95% CI: [{ci[0]:.2f}, {ci[1]:.2f}]')
```
